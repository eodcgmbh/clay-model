## Preparing the dataset
This notebook will create the image and label patches in order to use as input for the model.

By default we create chips of size `224 x 224` to feed them to the model.

Alternatively, you can run this in the console:
```bash
python claymodel/finetune/segment/preprocess_data.py data/cvpr/files data/cvpr/ny 224
```

Directory structure:
```
data/
└── cvpr/
    ├── files/
    │   ├── train/
    │   ├── val/
    │   └── test/
    └── ny/
        ├── train/
        │   ├── chips/
        │   └── labels/
        ├── val/
        │   ├── chips/
        │   └── labels/
        └── test/
            ├── chips/
            └── labels/
```

````python
import os
from pathlib import Path
import sys

def find_project_root(marker='claymodel'):
    current = Path.cwd()
    for path in [current] + list(current.parents):
        if (path / marker).exists():
            return path
    raise FileNotFoundError(f"Project root not found")

os.chdir(find_project_root())
print(f"Current working directory: {Path.cwd()}")
sys.path.append('.')
````

````python
INPUT_DIR = "data/GFM/files"
OUTPUT_DIR = "data/GFM/ny"
CHIP_SIZE = 224
````

````python
import subprocess

try:
    cmd = [
        "python",
        "claymodel/finetune/flood_detection/preprocess_data_gfm.py",
        INPUT_DIR,
        OUTPUT_DIR,
        str(CHIP_SIZE)
    ]
    
    print(f"Running preprocessing with:")
    print(f"Input directory: {INPUT_DIR}")
    print(f"Output directory: {OUTPUT_DIR}")
    print(f"Chip size: {CHIP_SIZE}")
    
    result = subprocess.run(
        cmd,
        check=True,
        capture_output=True,
        text=True
    )
    
    print("\nPreprocessing completed successfully!")
    if result.stdout:
        print("Output:", result.stdout)
        
except subprocess.CalledProcessError as e:
    print(f"Error during preprocessing:")
    print(f"Exit code: {e.returncode}")
    print(f"Error output: {e.stderr}")
except Exception as e:
    print(f"Unexpected error: {str(e)}")
````
